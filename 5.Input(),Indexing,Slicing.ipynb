# **Slicing**
 - To extract multiple characters
 - Syntax of slicing-->  variablename[starting index number(included) :ending index number(not included): steps taken(by default it is 1)]
 - if step is positive  go in positive direction (left to right) and if it is negative go in negative direction(right to left)

```python
a="Hello"

print(a[0:4:1]) # output=Hell
```

```python
print(a[0:4:2])  # output=Hl
```

```python
print(a[4:0:1])  #output= No error (empty string as output )(In slicing you will not get error)
```

```python
print(a[4:0:-1]) # Output=olle
```

```python
print(a[4:0:-2]) # Output=ol
```

```python
a[10:0:-1]  # Note if 10 is out of range then it will go till last as 9 th index
```

```python
a[10:-1:-1] 
```

```python
a[100:0:-1]  # Note if 100 is out of range then it will start form last index
```

```python
a[0:10000:1] # Note if 10000 is not there then it will go till last as 9999 th index that is poora string cover ho jaaega
```

```python
a[:5:1] # if step is positive and starting index not given then by default it is 0 and in case of negative step it is last index
```

```python
a[:5:-1]
```

```python
a[:-1:-1]
```

```python
a[:3:1]
```

```python
a[:3:-1]
```

```python
a[:4:-1]
```

```python
a[0::1] # steps =+ve ending point not given the ending point=len(string) (i.e move until the end of string in +ve direction)
```

```python
a[0::-1] # steps negative ending point not given then ending point=-1 hoga (i.e move until the end of string in -ve direction)
```

```python
a[0:5:] # when 3rd one not given it is by default 1 
```

```python
a[0::1] # No need to give the value of 2nd one if steps not given if you have to traverse till last
```

```python
# when 1 colon is given the first one will be  starting and the second part will be ending by default.
```
