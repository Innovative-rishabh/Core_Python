# Dictionary Methods

- dict.items()
  - items() method returns a list of key-value pairs in the dictionary.
  -  this will give list of tuples where in each tuple contains key-value pairs.

```python
marks={'history':45, 'Hindi':56}
marks
```

```python
print(marks.items()) # returns a list of key-value pairs in the dictionary.
```

**dict.keys()**
- keys method returns a list of keys in the dictionary

```python
marks={'history':45, 'Hindi':56}
marks.keys() #  returns a list of keys in the dictionary
```

**dict.values()**
- returns a list of values in the dictionary.

```python
marks.values()
```

**to add a single item into a dictionary**
- option1

```python
marks={'history':45, 'Hindi':56}
marks['english']=80 # if english naam ka key already available then it will replace the value otherwise add
print(marks)
```

```python
marks={'history':45, 'Hindi':56}
marks.update({'Sanskrit':90})
print(marks)
```

```python
marks={'history':45, 'Hindi':56}
marks.update({'history':90}) # if already present update it otherwise add it
print(marks)
```

```python
marks={'History':45, 'Hindi':56}
marks.update({'Science':98,'SST':99}) # adding multiple items in the dictionary
marks
```

### deleting item from a dictionary
 - option 1: del dictname[key]

```python
marks={'History':45, 'Hindi':56}
del marks['History'] # this will delete the value inside the key named as History so without any value key ka koi matlab nhi hota hence it also get deleted eventually
print(marks)
```

- option 2: dict.pop(key)

  
- pop() method removes the value of the specified key.
- if the key does not exist, it raises a key error.
  

```python
marks={'History':45, 'Hindi':56}
marks.pop("Hindi") # Hindi key ko delete kr diya bina key ke value ka kya kaaam hence value inside it also gets deleted eventually
print(marks)
```

### option 3: dict.clear()
- dict.clear() method removes all items of a dictionary .


```python
marks={'History':45, 'Hindi':56}
marks.clear()
print(marks)
```

### dict.copy()
- copy() method returns a shallow copy of the dictionary
- changes to the copy of the dictionary  do not affect the original dictionary.

```python
b={'history':45,'hindi': 56}
a=b.copy()
a.update({"maths":90})
print(a)
print(b)
```

### dict.get()
- get() method returns the value of the specified key
- if the key is not present inside the dictionary, it will return "None"
- we can get the value by using another way also dict_name[key],if the key is not present inside the dictionary, it will return "KeyError"

```python
b={'history':55,'hindi': 56}
b.get("history")
```

```python
print(b.get("Maths"))#  if the key is not present inside the dictionary, it will return "None"
```

### to replace the values in dictionary

```python
b={'history':55,'hindi': 56}
b["maths"]=b.pop("hindi") # go and see what actually pop does 
print(b)
```
