```python
# Write a program to collect 5 numbers from the user and store the odd no and even no  in a seperate- seperate  list.

even=[]
odd=[]
for i in range(5):
    n=eval(input("Enter the number from the user: "))
    if n%2!=0:
        odd.append(n)
    else:
        even.append(n)

print(odd)
print(even)
```

```python
# write a program to print even number sum and odd number sum in the given list

evensum,oddsum=0,0
for i in range(5):
    n=eval(input("Enter the number from the user: "))
    if n%2==0:
        evensum+=n
    else:
        oddsum+=n

print("The sum of even numbers in the list is: ",evensum)
print("The sum of odd numbers in the list is: ",oddsum)
```

```python
## VVI In Python, the else attached to a loop is not tied to the if, it’s tied to the loop’s completion status.
## 👉 Think of it like this:
## “Run else only if the loop did NOT exit early.”
## And the main way a loop exits early is with break.

l=[1,2,3,4]

for i in l:
    print(i)
    if i==3:
        break

else:
    print("Loop executed without any break ")
```

# Nested For Loop

```python
for i in range(1,3):
    for j in range(2,4):
        print(i,j)
    
```

```python
i=1
for j in range(2,4):
    print(i,j)
       
```

```python
i=2
for j in range(2,4): 
    print(i,j)
       
```

**break keyword within the nested for loop will break internal for loop**

```python
for i in range(2):
    for j in range(3,8):
        s=i+j
        if s==6:
            break
        print(s)
```

```python
i=1

for j in range(3,8):
    s=i+j
    if s==6:
        break
    print(s)
```
