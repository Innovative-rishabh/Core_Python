# Functions



- A function is a group of releated statements that performs a specific tasks.
- A function is a set of statements that take inputs,do some specific computation and produces output.
- It avoids repetition and makes code reusable.
- If we use functions written by others in the form of library, it can be termed as library function.
- No of arguments while defining the function is equal to the no. of arguments while calling the function(otherwise you will get error).



- Inbuilt functions = Integrated with python.
- User Defined Functions= The functions that we create.

  

````python
#  A function was created  for adding the numbers in a list.
def l_s(l):
    s = 0
    for i in l:
        s = s + i
    return s
    
````

````python
l=[1,2,3,4,5]
l_s(l)
````

### Default Arguments

While creating a function , we are assigning a default value to an argument by using the assignment operator(=)

- If you do not assign value for default arguments,it will automatically selects the default value.
- If you assign the value for default argument, it will consider that value which you have assigned.


````python
def greet(name,msg):
    print("Hello ",name,msg)

````

````python
greet("Rishabh","Good Morning")
````

````python
def greet(name,msg="Good Morning"):
    print("Hello ",name,msg)
````

````python
greet("Rishabh")
````

````python
greet("Rishabh","Sut Jaa!") # If you assign the value for default argument, it will consider that value which you have assigned.
````

### Arbitrary arguments
- Sometimes we even do not know how many arguments we are going to pass through the function in advance.
- Python allows to handle this situation with the help of arbitray number of arguments.
  

````python
def greet(*n):  # agar * ke sath arguments aaye to samajh jaana arbitrary arguments honge (unlimited number of arguments)
    for i in n:
        print("Hello",i)
    
    
````

````python
greet(1,2,3,4,"Rishabh") # All these arguments that are passed in as arbitrary arguments will be going to stored in a single tuple.
````

````python
def greet(n):  # agar *  hata de to function ke andar jo likha ha wo treat hoga as single arguments
    for i in n:
        print("Hello",i)
````

````python
greet((1,2,3,4,"Rishabh"))# Since n ek single argument hai isiiye tuple ke andar items ko pass kiya kyuki 1 tuple will be treated as 1 argument instead of single argument if we pass more than 1 argument it will throw positional argument error
````

````python
def count(l):
    m,f=0,0
    for i in l:
        if i=="M":
            m=m+1
        else:
            f=f+1

    return m,f
        
    
````
