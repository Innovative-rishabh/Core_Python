# if-block


- it checks the condition for 1 time.
- the if block will be executed , if the the condition is true and if the condition is False, it ignores the if block.

# Loop (Repetion of the block of code)

- While loop:
    - Till the condition is satisfied , repeat the block of code.
    - Till the condition is satisfied ,the loop executes.

- When While Loop Stops:
     - Condition  should  be false.
     - Or , Break should be executed
    
      

**Difference between if and While**

 - if ,Checks condition for 1 time  in contrast While, checks the condition for multiple times.
 - In both of them the block of code is executed only when the condition is true.
   

```python
x=1
if x<5:
    print("python")
    x=x+1;
```

```python
x=1
while x<5:
    print("python")
    x=x+1;

print(x)
```

```python
x=1
while x<=15:  # It executed till condition is False.
    print(x)
    x=x+1;
    
    
print("rishu")
```

```python
x=1
while x<=15:  # while loop stooped when break is executed.
    print(x)
    if x==5:
        break
    x=x+1;
    
    
print("rishu")
```

```python
x=1
while x<=2:  # No use of break since the loop termintated earlier(execution of break depends on the condition written inside it)
    print(x)
    if x==10:
        break
    x=x+1;
    
    
print("rishu")
```

```python
x=1
while True:
    x=x+10
    print(x)
    if x>50:
        break


    
```

```python
s=0
while True:
    a=eval(input("Enter the value: "))
    s=s+a
    print("Sum till now: ",s)
    if s>1000:
        break

print(s)
    
```

```python
# Ask user to enter 10 values and store them in a list.

print("Enter your values,once all entries are completed,please enter #")

l=[]

while True:
    i=input("Enter the values: ")
    if i=="#":
        break
    j=int(i)
    l.append(j)

print(l)
```

```python
# Ask the user to input 5 values,do the sum of the values.

x=1
s=0

while x<=5:
    l=eval(input("Enter the values: "))
    s=s+l
    x=x+1

print(s)
    
```

```python
# Ask the user to enter values,do the sum of the values.

print("Enter your values,once all entries are completed,please enter #")
s=0
while True:
    i=input("Enter the values: ") # Note eval only works with int and float.
    if i=="#":
        break

    s=s+int(i)

print("The total sum is: ",s)
```

```python
num = input("Enter your phone number: ")

# Convert input to integer (in case user enters it as string)
num = int(num)

while num >= 10:   # repeat until single digit
    temp = num
    digit_sum = 0
    
    while temp > 0:
        digit_sum += temp % 10
        temp = temp // 10
    
    num = digit_sum

print("Single digit sum:", num)
```
