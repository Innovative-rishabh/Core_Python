# Datatypes

- Data==value
- Based on value, we have different types of objects
- Each object is a datatype
- Example Based on students strength,we have differnt class room size

### Fundamental Datatypes(stores only 1 value in single object)
- 1 numeric value without decimal--> we are going to  store that value as int object/int datatype
- 1 numeric value wit decimal--> we are going to  store that value as float object/float datatype
- 1 non numeric value        --> we are going to store that value as string object/string datatype
- True or False as value     --> we are going to store that value as string object/string datatype

#### Note: All the fundamental data types are immutable.
- Immutable--> we can not modify the value in the original object.
- Mutable--> we can modify the value in the original object.
  -Ex of Mutable Objects= List

### Advanced Datatypes(can store more than 1 value in single object)
- If you want to store more than 1 value in a single object then you will need advance datatypes

  - List datatype
  - Tuple datatype
  - Set datatype
  - dictionary datatype

- Note: Object==Containers== Data structures 

## 1.Integer(int)

 - numerical value with no decimal.
 - can be neither negative nor positive.

```python
a=10  #assigning the value to a variable
print(a)
```

```python
type(a)   # to check the data type: use type() function
```

```python
id(a)   #it will give the address of object in ram(it is also a function)
```

```python
a=45  # it can not modify the original value but will delete the previous value of a first and then will create a new variable named as a 
print(a)
```

```python
b=45 # same value as a but still has different address or location in ram
id(b)
```

```python
# Mutable ojects example(list datatype)

l=[1,2,3,4]
print(l)
```

```python
id(l)
```

```python
l[0]=10 # modifying the value placed at 0 index in list l
print(l)
```

```python
id(l) # here you can see the value of l is changed still its address is same(mutable object)
```

```python
l.clear()  # it will delete all the values inside the list named as l.(clear is a function)
print(l)
```

# 4.String(str)

- 1 non numerical value except True or False comes under string.
- a combination or sequence of characters(alphabets,numbers,special characters).
- we can use single or double quotes or triple quotes to represent strings.


```python
myString1='Hello'
print(myString1)

myString2="Hello"
print(myString2)

myString3='''Hello'''
print(myString3)
```

```python
# multi line strings can be denoted using triple quotes, ''' or """

name1="""data
science"""
print(name1)

name2='''data
science'''
print(name2)
```

```python
var1=123
type(var1)
```

```python

var2="123"   # numeric value isnide inside inverted comma is a string not integer
type(var2)
```

```python

var3="Rishabh 1" # note blank space is a character
len(var3)      # len is a function that is applicable o string that will give you as count of characters in the given string

#Note len function is not applicable on int or float or boolean,it will throw error
```

```python
a=123456789
# now you  want to calcualte the total no. of digits in the variable a 
# first convert int to string (type casting) and then apply len function
str(a) #typecasting
```

```python
len(str(a))
```

# String Concatenation
 - Joining of 2 or more strings into a single string is called concatenation
 - The "+" operator does this in python.Simply writing 2 string literals also can do the job.
 - Note : for using "+" operator for string concatenation both of them must be string only.

```python
"hello" + "world"
```

```python
"hello "+"world" 
```

```python
 "hello"+" "+"world"
```

```python
"siva"+2 # it will throw error  since  2 is not a string and string can't bwe conacatenated with int
```

```python
"siva" + "2" # No error since both of them are string here
```
