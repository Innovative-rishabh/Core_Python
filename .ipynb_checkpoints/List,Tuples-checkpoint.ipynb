# **List**
- It is one of the most used datatypes of python.
- It is an advance datatype where it can store multiple  values.
- Each value inside the list is called an item.
- A list can have 0 items to multiple items.
- Declaring a list items seperated by commas are enclosed within the brackets[].
- All the items in the list do not need to be of same type(hetrogeneous).
- List is an ordered sequence of items.

```python
# Empty List -> List with 0 items is called an empty list.
# object matlab room
# values matlab student inside the room
```
