- ### set is an unordered collection of items.
- ### set is defined by values seperated by comma inside braces.
- ### Set can store only immutable items(i.e it can not store list,set )
- ### since set is also mutable hence a set can not be placed inside another set.(hnce there is no any concept of nested sets)

```python
s={10,20,30,40,50}
print(s) # unordred order will be printed
```

```python
s={1,1.4,True,"srk",(10,20)}
print(s) # Set can store only immutable items while printing but order can be random
```

```python
s={1,1.4,True,"srk",(10,20),[1,2]}
print(s) # Set can store only immutable items(i.e it can not store list ) hence error
```

```python
s={1,1.4,True,"srk",(10,20),{1,2}} #since set is also mutable hence a set can not be placed inside another set.
print(s)
```

- ### Every items is unique inside sets(no duplicates)
- ### while creating the object it will remove the duplicates at the time of creation only.

```python
s={1,2,2,2,3,3,4}
len(s) # this will print the no. of unique values in the set since set does not contain duplicates.
```

- ### set does not support indexing.

```python
s={1,2,2,2,3,3,4}
print(s[-1]) # set' object is not subscriptable
```

# Set Methods

   ## set.add()
   - add method adds an element to the set
   - if the element is already present inside the set nothing will happen.

```python
s={1,2,3}
s.add(4)
s
```

```python
s={1,2,3}
s.add(2) # if the element is already present inside the set nothing will happen.
s
```

## se.update()
 - If you wnat to add multiple values in a set take them in a list or in a tuple and use update method to add them.
 - update method updates the set with another set or any iterable.

```python
s1={1,2,3}
s1.update([4,5,6]) #  - If you wnat to add multiple values in a set take them in a list and use update method to add them.
s1
```

```python
s={1,2,3}
s.add((4,5)) # add function will treat (4,5) as single value
s
```

```python
s={1,2,3}
s.update((4,5)) # update function will treat 4 and 5 as sperate values
s
```

### set.discard()
- it will remove the spcified element from the set.
- if the element does not exist in the set , it will not do anything.

```python
s={1,2,3,4,5}
s.discard(5)
s
```

```python
s={1,2,3,4,5}
s.discard(6)
s
```

### set.remove()
- it will remove the spcified element from the set
- if the element does not exist in the set , it will raise a key error.

```python
s={1,2,3,4,5}
s.remove(5)
s
```

```python
s={1,2,3,4,5}
s.remove(6)
s
```

### s.clear()
- clear method will remove all the elements from the set and will give a empty set.

```python
s={1,2,3,4,5}
s.clear()
s
```

```python
s={1,2,3,4,5}
del s     # delete the variable(object) hence NameError
s
```

### set,copy()
- copy method returns a shallow copy of the set
- changes to the copied set does not affect the original set.

```python
s1={1,2,3}
s2=s1.copy()
print(s1,id(s1))

print(s2,id(s2))
```

### set.union()
 - this method returns a new set with elements from both sets 

```python
setA={1,2,3}
setB={3,4,5}
setA.union(setB)

```

### set.intersection()
 - this method will return a new set with elements that are common to both the sets.
 - if no any common element present it will return empty set
   

```python
setA={1,2,3}
setB={3,4,5}
setA.intersection(setB)
```

```python
setA={1,2,3}
setB={4,5} # if no any common element present it will return empty set
setA.intersection(setB)
```

### set.isdisjoint()
 - isdisjoint() method will return true if no any common element present in bot the sets otherwise false

```python
setA={1,2,3}
setB={3,4,5}
setA.isdisjoint(setB)
```

```python
setA={1,2,3}
setB={4,5}
setA.isdisjoint(setB)
```

 ### set differnce()
 - difference method returns a new set with  elements in the set that are not present in other specified set.

```python
setA={1,2,3}
setB={2,3,4}
print(setA-setB) # set A ka wo element jo set B ka part na ho
print(setB-setA) # set B ka wo element jo set A ka part na ho
```

### set.symmetric_differnce()
- this method will return a new set that contains elemnts that are part of either of the sets but not both of them.
- it will give: (A union B)-(A intersection B)
  

```python
setA={1,2,3,"rishu"}
setB={2,3,4}

setA.symmetric_difference(setB) # it will give: (A union B)-(A intersection B)
```

### issubset()
- issubset method returns true if all elements of first set  are available in second set , otherwise returns false
- then a is called as subset and b is called as superset
  

```python
A={3,4,5,7,6}
B={3,4,5,6}
ans= A.issubset(B)
print(ans)

ans= B.issubset(A)
print(ans)
```

### issuperset()
- if all the elements of second set are available in the first set it will return true otherwise false.

```python
A={3,4,5,7,6}
B={3,4,5,6}
ans= A.issuperset(B)
print(ans)

ans= B.issuperset(A)
print(ans)
```
