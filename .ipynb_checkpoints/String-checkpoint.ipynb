# String Methods
- Note: String is an immutable object.

  - str.upper()
     - returns a copy of string with all characters in uppercase.
       
  

```python
s="True...123"
s.upper()
```

#### str.lower()
 - returns a copy of string with all characters in lowercase.

```python
s="True...123"
s.lower()
```

```python
s="HELLO WORLD"
print(s)
print(id(s))
s=s.lower()
print(s)
print(id(s))
```

#### str.swapcase()
- returns a copy of string with all uppercase convreerted lower case and vice versa


```python
s="True...123"
s.swapcase()
```

#### str.capitalize()
- returns a copy of string with the first letter being capitalized and the remaining letter as it is.
- if the 1st letter is number or it is in capiatal already then it will not change  anything.

```python
s="hello World"
s.capitalize()
```

#### str.title()
- returns a copy of string with first letter in each word is capitalized.

```python
s="gooD moRNing"
s.title()
```

#### str.strip()
- returns a copy of the string with leading and trailing whitespaces removed.

```python
s="  Python is easy  "
s.strip()
```

#### str.lstrip()   -> Left side whitesapces
- returns a copy of the string with leading whitespaces removed.

```python
s="  Python is easy  "
s.lstrip()
```

#### str.lstrip()¶ -> Right side whitesapces
- returns a copy of the string with trailing whitespaces removed.

```python
s="  Python is easy  "
s.rstrip()
```

#### str.replace()
- returns a copy of the string with all the occurences of specifid substring replaced with another string

```python
s="Good Morning"
print(s)
print(id(s))

s=s.replace("Good", "Bad")
print(s)
print(id(s))
```

#### str.isupper()
- returns a boolean stating whteher a string is in uppercase.
- It returns True if all cased characters in the string are uppercase AND there is at least one cased character.
- Only letters (A–Z, a–z) are considered cased characters.
- Uppercase letters → checked ✔️
- Lowercase letters → checked ✔️
- Digits (1,2,3) → ignored ❌
- Spaces → ignored ❌

```python
s="SIVA"
s.isupper()
```

```python
s="siva"
s.isupper()
```

```python
s="Siva"
s.isupper()
```

```python
s="SIVA "
s.isupper()# every text is in uppercase
```

```python
s="SIVA 123"
s.isupper()
```
