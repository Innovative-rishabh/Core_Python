**Types of operators**
 - Arithmetic operators
 - Assignment operators
 - Comparison(Releational) operators
 - Logical (Boolean) operators
 - Membership operators
 - Identity operators

**Assignment Operators**
- They are used in python to assign the values to the variables.
- (=,+=,-=,*=,/=,%=,//=,**=) are assignment operators.
- in above operators first right side part  will be executed and then the value will be assigned to left side variables.
- Paranthesis () is also an assignment operator.

```python
x=5
print(x)
```

```python
x=5
x=x+10
print(x)
```

```python
x=5
x+=10 # same as x=x+10
print(x)
```

```python
x=5
x*=10 # same as x=x*10
print(x)
```

```python
x=5
x/=10 # same as x=x/10
print(x)
```

 **Logical Operators/Boolean Operators**
 - Logical operators are used to check multiple conditions at a time.
 - It returns bool type only.
 - and,or,not are logical operators.
 - when we have "or" operator then at least 1 condition should be True then only the overall output will be True.
 - when we have "and" operator then all the condition should be True then only the overall output will be True.

**Logical Operators Precedence**

- Logical NOT
- Logical AND
- Logical OR

```python
5>3 and  2>3 # Note among comparison and logical operators the precedence of comparison operator is higher.
```

```python
5>3 or 5<3
```

```python
not False # opposite to False 
```

```python
not True # opposite to True 
```

```python
5>3 or 2>3 and 5>3 # True or False and True ( The precedence of logical opertaors are 1. not 2. and 3. or)
```

```python
False or not(5<2) and True
```

**Identity Operators**

- They are used to check two values or variables are indicating to same object or not.
- It returns output as boolean.
- is, is not are identity operators in Python.
- is operator(# is-True  if the operands are identical).
- is not operator(# is not-True  if the operands are  not identical).

```python
a=5
b=5

print(a is b) # id(a)==id(b)
```

```python
s1="Satish"
s2="satish"
print(s1 is s2) # id(s1)==id(s2)
```

```python
s1="satish"
s2="satish"
print(s1 is s2) # id(s1)==id(s2)
```

```python
a=6
b=8

a is not b
```

```python
a=[1,2,3]
b=a
print(a is b) # deep copy refers to same object
```

```python
a=[1,2,3,4]
b=[1,2,3,4] # since b  is not a deep copy hence the ids of both the list will be different
print(a is b)
```

**Membership Operator**

- It is used to test whether a value or variable is found in a sequence (string,list,tuple,set and dictionary).
- It returns output as boolean.
- in, not in are the membership operators.

```python
a="venkatesh"
"esh" in a
```

```python
lst=[1,2,3,4]

5 in lst
```

```python
a=[1,2,3,4,5,6]

9 not in a
```

**Only Arithmetic operators returns values.**

**Remaining Operators , return boolean values.**

- Membership Operators

```python
b= (1>2) or (3>=3) and (5<4)
print(b)
```
