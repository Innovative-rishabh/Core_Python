# Random Module

- this module defines several functions to generate random numbers
- we can use these functions while developing games,in cryptography and in authentication process to generate random numbers

```python
import random
```

```python
random.random() # random module ke andar random function was created to generate some float value between [0,1) i.e 0 included and 1 not included
```

```python
# now if you want to print 5 random values between [0,1) --> use for loop
```

```python
for i in range(5):
    print(random.random())
```

```python
# randint()--> It is a Function inside random module to generate random integers(not a float value) between 2 given numbers(both inclusive)
```

```python
random.randint(1,100)
```

```python
for i in range(5):
    print(random.randint(1,100))
```

```python
# randrange()--> It is a Function inside random module to generate random integers between 2 given numbers(both inclusive) and also the 3rd argument inside it shows step
```

```python
random.randrange(2,50) # if step not provided it will work in same way as of randint step value by default 1 hota hai!
```

```python
random.randrange(2,50,3) # random.randrange(2, 50, 3) generates a random number starting from 2, less than 50, with a step of 3.
# So the possible outputs are: 2, 5, 8, 11, 14, 17, 20, 23, 26, 29, 32, 35, 38, 41, 44, 47
```

```python
random.randrange(2,50,2) 
```

```python
# choice()=> it is a function inside random module that does not returns number instead it returns random object from the given sequence(list,tuple or string)
list=["sunny","bunny","chinny","vinny"]
random.choice(list)
```
