**loop** ->repeating the code


**while** -> till condition satisfies

**while loop** ->repeat the code till the condition is satisfied.

**for**-> for each value in sequence

**for loop**->for each value in sequence repeat the block of code

**sequence**-> str,list,tuple,set,dict,range 

**Syntax of for loop**

- for variable_name in sequence:

               block of code

## What is the difference between if block and a loop(for/while)?
## If block checks the condition for 1 time only on the other hand a loop(while/for) checks the condition multiple times.

### while loop

```python
l=[10,"a",20,30,40]
i=0
print(len(l))
while(i<len(l)):
    print(l[i])
    i=i+1
```

### for loop

```python
l=[10,"a",20,30,40]

for i in l: # here sequence is a list named l
    print(i)
```

```python
st="Rishabh"

for i in st:
    print(i)
```

```python
l=[1,2,3,[4,5]]

for i in l: # here sequence is a list named l
    print(i)
```

```python
l=(1,2,3,[4,5])

for i in l: # here sequence is a tuple named l
    print(i)
```

```python
l={"a",2,3,"b",5,6,"abc",(1,2,3)}

for i in l: # sequence is a set (order is not important while printing)
    print(i)
```

```python
s={1,1,2,2,1,2,3,3,4}
for i in s: # since set does not store duplicates
    print(i)
```
