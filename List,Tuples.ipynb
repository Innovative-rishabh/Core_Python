## Fundamental Data Types
 - They can store only 1 value inside the object.
 - All fundamental datatypes are immutable(we can not modify the values inside or within the same object).
 - Eg= int,float,string,bool

## Advanced Data Types
 - They can store more than 1 value inside the object.
 - Eg= List,Tuple,Set,Dictionary
 - list,set and dictionary are mutable(we can modify the values within the same object).
 - while tuple is immutable.

```python
 # shallow copy= It indicates to different objects
a=[1,2,3,4]
b=a.copy()
b.append(5)
print(a)
print(b)
print(id(a),id(b))
```
