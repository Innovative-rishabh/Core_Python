# Decision Making Statements
- It checks the condition everytime and if the condition is true it will run otherwise not.
- These are also called as Conditional Statements or Control Flow statements in python.
   - if
   - if-else
   - if-elif
   - if-elif-else
   - nested if

# Block Structure

- The code that is executed when a specific condition met is defined in a "block".
- Statements preceding blocks generally end with a colon(:).
- In Python, the block structure is signalled by changes in indentation.
- Each line of code in a certain block level must be indented equally and indented more than the surrounding scope.

# if-else

- If the condition written inside if block is failed then automatticaly the statements inside else block will be executed.

```python
a,b,c,d=2,2,3,4

if (a==b and c!=d):
    print("abc")

else:
    print("def")
```

```python
a,b,c,d=2,2,3,3

if (a==b and c!=d):
    print("abc")

else:
    print("def")
```

# if-elif

- if condition true -> do not check below in elif
- if the condition in if block failed then check the condition in elif block

```python
if 1<5:
    a=50
    print(a+10)

elif 2<3:          # above condition is not failed hence it will not check down in elif
    print("data")

elif 2<5:
    print("Python")
```

```python
if 1>5:
    a=50
    print(a+10)

elif 2>3:
    print("data")

elif 2<5:
    print("python")
```

```python
if 1>5:
    a=50
    print(a+10)

elif 2<3:
    print("data")

elif 2<5:
    print("python")
```

# Rule: In an if-elif-elif-else chain, only the first condition that is True gets executed.

```python
marks=eval(input("Enter your Marks"))

if marks>90:
    print("Grade A")

elif marks>=70:
    print("Grade B")

else:                 # if all the above conditions are failed then else block will be executed automatically.
    print("Grade F")

    
   
```

# Nested-if
- We can have an if elif else statements inside another if elif else statement.This is called as nested if statements.


```python
num=2

if num>0:
    if num==0:
        print("Zero")
    else:
        print("Positive Number")
else:
    print("Negative Number")
```
